## Reading The Features

Looking at the data 2 main things stand out, person_gender & loan_int_rate. 
### Gender

Banks don't use gender as a metric to determine if someone could get a loan. Could it improve the performance of the model? Maybe but we should remove it because banks don't use that as criteria. We could just be seeing a pattern that adds noise.

### Interest Rate

If we include interest rate it would be too easy for the model to figure out who gets qualified and who doesnt. We should investigate this because banks normally set this after approval. However, In many datasets the interest rate is based on risk so it may still be valid.

### Target Variable

Our target variable will be loan_status. That tells us if someone got approved or not (0 = not approved, 1 = approved)

# Reading the df description

Looking at the description we can see that there is someone who is 144 years old. That is highly unlikely so we should clean that up by caping those values. We can set all the ages over 90 to 90

```python
# get ages and credit history length for people over 90
over_90_data = df.loc[df['person_age'] >= 90, ['person_age', 'cb_person_cred_hist_length']].values
over_90_data
```

Based off of these results we can see there are only 8 entries above the age of 90 and if we cap the age at 90 the credit history length wouldn't become contradictory. Therefore, we can proceed and cap the age at 90.

```python
df_capped = df.copy()
df_capped['person_age'] = np.where(df_capped['person_age'] > 84, 84, df_capped['person_age'])
df_capped.describe()
```

```python
df_cleaned = df.copy()

df_cleaned['person_age'] = df["person_age"].clip(upper = 90)
print(df_cleaned['person_age'].max())

#removes person gender from the df as the cleaned product
df_cleaned = df_cleaned.drop(columns=['person_gender'])
display(df_cleaned.columns)


x = df_cleaned.drop(columns=['loan_status']) #all columns but loan status (training data for the model)
y = df_cleaned['loan_status'] #only loan status column (answesrs for if they are appoved or not)


display(x.head())
display(y.head())
```

```python
# Get unique entries for the specified columns
print("Finding Unique Values for Categorical Columns:")
unique_education = df['person_education'].unique()
unique_home_ownership = df['person_home_ownership'].unique()
unique_loan_intent = df['loan_intent'].unique()

print("Unique person_education:", unique_education)
print("Unique person_home_ownership:", unique_home_ownership)
print("Unique loan_intent:", unique_loan_intent)

# Find rows with any negative values in numerical columns
print("\n\nChecking for negative values in numerical columns:")
numerical_cols = ['person_age', 'person_income', 'person_emp_exp', 'loan_amnt', 'loan_int_rate', 'loan_percent_income', 'cb_person_cred_hist_length', 'credit_score', 'loan_status']
negative_rows = df[(df[numerical_cols] < 0).any(axis=1)]
print(negative_rows.shape[0], "rows with negative values found.")

#get some statistics on the "other" home ownership group to determine if we should drop it or not
print("\n\nStatistics for 'OTHER' home ownership group:")
display(df["person_home_ownership"].value_counts())
other_home_ownership = df[df['person_home_ownership'] == 'OTHER'][['person_age', 'person_income', 'loan_intent']]
display(other_home_ownership)
other_home_ownership.describe()
```

Looking at the categorical columns we can see that all of them are valid and don't need to be dropped. We also checked if there are any negative values in the numerical columns which there isn't so we're good on that front. Now we just need to map the categroical columns.

# Mapping Categorical Variables

```python
education_map = {
    "High School": 1,
    "Associate": 2,
    "Bachelor": 3,
    "Master": 4,
    "Doctorate": 5,
}

if "person_education" in df_cleaned.columns:
    df_cleaned["person_education"] = df_cleaned["person_education"].map(education_map)

nominal_cols = ["person_home_ownership", "loan_intent"]


if nominal_cols:
    df_cleaned = pd.get_dummies(df_cleaned, columns=nominal_cols, drop_first=True, dtype=int)

display(df.head())
display(df_cleaned.head())
```
